```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/drive/MyDrive/Pindahan Dokumen/ktm_exc_scan.jpeg")
cv2.rectangle(img,pt1=(1560,30),pt2=(1100,200),color=(0,255,0),thickness=(10))
cv2.rectangle(img,pt1=(200,200),pt2=(300,300),color=(0,0,255),thickness=(15))
cv2.circle(img,center=(100,100),radius=50,color=(255,0,0),thickness=(8))
cv2.circle(img,center=(400,400),radius=10,color=(255,0,0),thickness=(100))
cv2.line(img,pt1=(0,0),pt2=(509,509),color=(255,255,255),thickness=(5))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```
